# C++20 (`g++` 整格编译内核) 实验演示

- **支持标准 `int main()`**：每个单元格独立调用 `g++ -std=c++20 -O0` 编译运行，反复修改重跑**不会出现重定义（Redefinition）错误**。
- **支持 `// @input:` 测试输入**：含有 `cin` 的程序可在代码顶部写 `// @input: 测试数据` 自动注入标准输入，也可直接运行后在弹窗中输入。
- **报错位置源码对照**：编译报错返回标准 `CompileError` 并附带报错行号上下文；运行崩溃返回 `RuntimeSignalError` / `ExitCode`。

```cpp
// @input: 5
// @input: 10 20 30 40 50
#include <iostream>
#include <vector>
using namespace std;

int main() {
    int n;
    if (!(cin >> n)) return 0;
    vector<int> nums(n);
    int sum = 0;
    for (int i = 0; i < n; ++i) {
        cin >> nums[i];
        sum += nums[i];
    }
    cout << "元素个数: " << n << ", 总和: " << sum << endl;
    return 0;
}
```

```cpp
// 演示故意写错语法（缺少分号），运行会触发 CompileError 并输出报错行号对照；
// 补上分号后直接再次运行本单元格即可通过，无需重启内核。
#include <iostream>
using namespace std;

int main() {
    int a = 42
    cout << "修复成功: a = " << a << endl;
    return 0;
}
```
